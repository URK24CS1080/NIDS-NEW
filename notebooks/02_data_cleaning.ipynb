```python
all_labels = set()

for file in os.listdir(DATA_DIR):

    if file.endswith(".csv"):

        file_path = os.path.join(DATA_DIR, file)

        temp_df = pd.read_csv(file_path)

        # Clean column names
        temp_df.columns = temp_df.columns.str.strip()

        # Get labels
        labels = temp_df["Label"].dropna().unique()

        all_labels.update(labels)

print("All labels found in the dataset:")
for label in sorted(all_labels):
    print(label)

print("\nTotal unique labels:", len(all_labels))
```

```python
def clean_dataframe(df):
    
    # 1. Clean column names
    df.columns = df.columns.str.strip()
    
    # 2. Replace infinite values with NaN
    numeric_columns = df.select_dtypes(include=np.number).columns
    
    df[numeric_columns] = df[numeric_columns].replace(
        [np.inf, -np.inf],
        np.nan
    )
    
    # 3. Fill numeric missing values with median
    for column in numeric_columns:
        if df[column].isnull().sum() > 0:
            df[column] = df[column].fillna(df[column].median())
    
    # 4. Clean attack label encoding
    df["Label"] = df["Label"].str.replace(
        "Web Attack �",
        "Web Attack -",
        regex=False
    )
    
    return df


print("Cleaning function created successfully.")
```

```python
test_df = pd.read_csv(
    os.path.join(
        DATA_DIR,
        "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv"
    )
)

print("Before cleaning:")
print("Missing values:", test_df.isnull().sum().sum())

numeric_test = test_df.select_dtypes(include=np.number)
print("Infinite values:", np.isinf(numeric_test).sum().sum())

print("\nLabels before cleaning:")
print(test_df[" Label"].unique())

# Apply cleaning
test_df = clean_dataframe(test_df)

print("\n" + "=" * 50)
print("After cleaning:")

print("Missing values:", test_df.isnull().sum().sum())

numeric_test = test_df.select_dtypes(include=np.number)
print("Infinite values:", np.isinf(numeric_test).sum().sum())

print("\nLabels after cleaning:")
print(test_df["Label"].unique())
```

```python
cleaned_data = []

for file in sorted(os.listdir(DATA_DIR)):

    if file.endswith(".csv"):

        file_path = os.path.join(DATA_DIR, file)

        print(f"Processing: {file}")

        # Load file
        temp_df = pd.read_csv(file_path)

        # Clean file
        temp_df = clean_dataframe(temp_df)

        # Store cleaned file
        cleaned_data.append(temp_df)

        print(f"Rows: {len(temp_df):,}")
        print("Done.\n")


print("=" * 80)
print("Combining all cleaned files...")

cleaned_df = pd.concat(
    cleaned_data,
    ignore_index=True
)

print("\nCleaning completed.")
print("Total rows:", f"{len(cleaned_df):,}")
print("Total columns:", len(cleaned_df.columns))
```

```python
print("=" * 80)
print("FINAL CLEANING VERIFICATION")
print("=" * 80)

# Missing values
total_missing = cleaned_df.isnull().sum().sum()

# Infinite values
numeric_cleaned = cleaned_df.select_dtypes(include=np.number)
total_infinite = np.isinf(numeric_cleaned).sum().sum()

# Duplicate rows
total_duplicates = cleaned_df.duplicated().sum()

# Unique labels
unique_labels = cleaned_df["Label"].unique()

print("Total rows:", f"{len(cleaned_df):,}")
print("Total columns:", len(cleaned_df.columns))

print("\nMissing values:", total_missing)
print("Infinite values:", total_infinite)
print("Duplicate rows:", total_duplicates)

print("\nUnique labels:")
for label in sorted(unique_labels):
    print("-", label)

print("\nTotal unique labels:", len(unique_labels))
```

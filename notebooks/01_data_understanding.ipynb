```python
import os
import pandas as pd

DATA_DIR = "../data/raw/CIC-IDS2017/MachineLearningCVE"

print("Dataset directory:")
print(os.path.abspath(DATA_DIR))

print("\nFiles found:")

for file in os.listdir(DATA_DIR):
    print(file)
```

```python
file_path = os.path.join(
    DATA_DIR,
    "Monday-WorkingHours.pcap_ISCX.csv"
)

df = pd.read_csv(file_path)

print("Shape:", df.shape)

df.head()
```

```python
print("Number of columns:", len(df.columns))

for i, column in enumerate(df.columns):
    print(i, "->", repr(column))
```

```python
print("Unique labels:")
print(df[" Label"].unique())

print("\nNumber of unique labels:")
print(df[" Label"].nunique())
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Memory usage:", df.memory_usage(deep=True).sum() / (1024**2), "MB")
```

```python
for file in os.listdir(DATA_DIR):
    
    if file.endswith(".csv"):
        
        file_path = os.path.join(DATA_DIR, file)
        
        labels = pd.read_csv(
            file_path,
            usecols=[" Label"]
        )
        
        print("\n" + "=" * 70)
        print(file)
        print("Rows:", len(labels))
        print("Unique labels:", labels[" Label"].nunique())
        print(labels[" Label"].value_counts())
```

```python
total_rows = 0

for file in os.listdir(DATA_DIR):
    
    if file.endswith(".csv"):
        
        file_path = os.path.join(DATA_DIR, file)
        
        rows = sum(1 for _ in open(file_path, encoding="utf-8", errors="ignore")) - 1
        
        print(f"{file}: {rows:,} rows")
        
        total_rows += rows

print("\n" + "=" * 60)
print(f"Total rows across all files: {total_rows:,}")
```
